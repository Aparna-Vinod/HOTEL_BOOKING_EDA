#Data cleaning

```python
# Create a cleaned copy of the original dataset

df_clean = df.drop_duplicates().copy()

print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", df_clean.shape)

print("Duplicate Records Remaining:",
      df_clean.duplicated().sum())
```

```python
# Handle missing values

# Fill missing children with the median
df_clean['children'] = df_clean['children'].fillna(
    df_clean['children'].median()
)

# Fill missing country values
df_clean['country'] = df_clean['country'].fillna('Unknown')

print("Missing Values After Treatment:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
# Convert reservation date into datetime format

df_clean['reservation_status_date'] = pd.to_datetime(
    df_clean['reservation_status_date'],
    errors='coerce'
)

# Convert agent and company identifiers to nullable integers
df_clean['agent'] = df_clean['agent'].astype('Int64')
df_clean['company'] = df_clean['company'].astype('Int64')

print(df_clean.dtypes)
```

```python
# Calculate total guests

df_clean['total_guests'] = (
    df_clean['adults'] +
    df_clean['children'] +
    df_clean['babies']
)

# Calculate total nights stayed

df_clean['total_nights'] = (
    df_clean['stays_in_weekend_nights'] +
    df_clean['stays_in_week_nights']
)

df_clean[['adults', 'children', 'babies',
          'total_guests', 'total_nights']].head()
```

### Data Cleaning and Preparation – Observations

The original dataset contained 119,390 records and 32 columns. During data cleaning, 31,994 exact duplicate records were removed, resulting in 87,396 records.

Missing values in the children column were replaced with the median value, while missing country values were labelled as 'Unknown'. Missing agent and company identifiers were retained because their actual values could not be reliably determined.

The reservation status date column was converted into datetime format. Additionally, two new features, total_guests and total_nights, were created to support further analysis.

After cleaning, the dataset contained no exact duplicate records or missing values in the columns other than agent and company. Unusual numerical values were identified for further investigation before visualisation.

#Visualization 1 – Hotel Type

Question: Which type of hotel receives the highest number of bookings?

Interpretation: City Hotels received significantly more bookings than Resort Hotels, indicating that they account for a larger share of total reservations.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(data=df_clean, x='hotel')

plt.title('Booking Distribution by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Number of Bookings')

plt.tight_layout()
plt.show()
```

#Visualization 2 – Cancellation Rate

Question: Which hotel type has a higher booking cancellation rate?

Interpretation: City Hotels have a higher cancellation rate (approximately 30%) than Resort Hotels (approximately 23%), indicating more frequent cancellations among City Hotel bookings.

```python
cancellation_rate = (
    df_clean.groupby('hotel')['is_canceled']
    .mean()
    .mul(100)
    .reset_index(name='cancellation_rate')
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=cancellation_rate,
    x='hotel',
    y='cancellation_rate'
)

plt.title('Booking Cancellation Rate by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Cancellation Rate (%)')

plt.tight_layout()
plt.show()
```

#Visualization 3 – Monthly Booking Trends

Question: During which months do hotels receive the highest number of bookings?

Interpretation: August recorded the highest number of bookings, followed by July. Booking activity was comparatively lower during November and December.

```python
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_bookings = (
    df_clean['arrival_date_month']
    .value_counts()
    .reindex(month_order)
    .rename_axis('arrival_date_month')
    .reset_index(name='booking_count')
)

plt.figure(figsize=(12, 5))

sns.barplot(
    data=monthly_bookings,
    x='arrival_date_month',
    y='booking_count'
)

plt.title('Monthly Distribution of Hotel Bookings')
plt.xlabel('Arrival Month')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

#Visualization 4 – Lead Time

Question: How far in advance do customers generally make reservations?

Interpretation: The distribution is strongly right-skewed, indicating that many customers book closer to their arrival dates, while fewer make reservations far in advance.

```python
# Use bookings with lead time up to 365 days for a clearer chart
lead_time_data = df_clean[
    df_clean['lead_time'] <= 365
]

plt.figure(figsize=(10, 5))

sns.histplot(
    data=lead_time_data,
    x='lead_time',
    bins=40,
    kde=True
)

plt.title('Distribution of Hotel Booking Lead Time')
plt.xlabel('Lead Time (Days)')
plt.ylabel('Number of Bookings')

plt.tight_layout()
plt.show()
```

**Note:** For better visualization, bookings with a lead time above 365 days (565 records) were excluded from this chart. This filtering was applied only for visualization purposes; the original dataset remains unchanged.

#Visualization 5 – ADR Distribution

Question: How are average daily rates distributed across bookings?

Interpretation: Most ADR values are concentrated between approximately 50 and 150, with the highest frequency around 90–100. The right-skewed distribution indicates fewer bookings at higher daily rates.

```python
# Remove negative ADR values for this specific analysis
adr_analysis = df_clean[df_clean['adr'] >= 0].copy()

# Calculate the 99th percentile
adr_99 = adr_analysis['adr'].quantile(0.99)

# Limit displayed values to the 99th percentile
adr_plot = adr_analysis[
    adr_analysis['adr'] <= adr_99
]

plt.figure(figsize=(10, 5))

sns.histplot(
    data=adr_plot,
    x='adr',
    bins=40,
    kde=True
)

plt.title('Distribution of Average Daily Rate (ADR)')
plt.xlabel('Average Daily Rate')
plt.ylabel('Number of Bookings')

plt.tight_layout()
plt.show()
```

**Note:** Negative ADR values and values above the 99th percentile (261.406) were excluded from this chart to improve readability. This filtering was applied only for visualization purposes; the original dataset remains unchanged.

#Visualization 6 – Market Segment

**Question:** Which market segment contributes the highest number of bookings?

**Interpretation:** Online TA records the highest number of bookings, followed by Offline TA/TO and Direct. This indicates that online travel agents are the most common booking channel in the dataset.

```python
plt.figure(figsize=(10, 5))

market_order = df_clean['market_segment'].value_counts().index

sns.countplot(
    data=df_clean,
    x='market_segment',
    order=market_order
)

plt.title('Distribution of Bookings by Market Segment')
plt.xlabel('Market Segment')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=35)

plt.tight_layout()
plt.show()
```

#Visualization 7 – Correlation Heatmap

What relationships exist between the numerical variables?

The heatmap shows positive relationships between stay-night variables and total nights, as well as between guest counts and total guests. It also helps identify associations between booking characteristics and cancellation status. Correlation does not necessarily imply causation.

```python
correlation_columns = [
    'lead_time',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'adults',
    'children',
    'babies',
    'previous_cancellations',
    'booking_changes',
    'adr',
    'total_of_special_requests',
    'total_guests',
    'total_nights',
    'is_canceled'
]

correlation_matrix = df_clean[
    correlation_columns
].corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')

plt.tight_layout()
plt.show()
```
